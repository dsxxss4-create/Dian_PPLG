Tuga_Ui_MUHAMMAD ADIKAMEDIAN PUTRA

```python
import ipywidgets as widgets
from IPython.display import display


# membuat input nama pemain
input_nama = widgets.Text(
    value='',
    placeholder='Ketik nama pemain',
    description='Nama:',
    disabled=False
)

# Membuat pilihan level menggunakan slider
input_level = widgets.IntSlider(
    min=1,
    max=100,
    description='Level:'
)

# Membuat pilihan role menggunakan dropdown
input_role = widgets.Dropdown(
    options=['Warrior', 'Mage', 'Archer'],
    description='Role:'
)

# Membuat tombol daftar
tombol_daftar = widgets.Button(
    description='Daftar',
    button_style='primary'
)

# Tempat menampilkan hasil (output)
output_hasil = widgets.Output()



# Membuat fungsi logika saat tombol diklik
def daftar_klik(b):
    with output_hasil:
        output_hasil.clear_output() # Membersihkan teks cetakan sebelumnya
        print(f"Pemain {input_nama.value} dengan role {input_role.value} level {input_level.value} telah siap bertarung!")

# Menyambungkan tombol dengan fungsi di atas
tombol_daftar.on_click(daftar_klik)


# Menampilkan semua widget form ke layar
display(input_nama, input_level, input_role, tombol_daftar, output_hasil)
```
